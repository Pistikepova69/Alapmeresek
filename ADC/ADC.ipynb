```python
feszultsegek = [0.00, 0.20, 0.40, 0.61, 0.82, 1.01, 1.20, 1.42, 1.60, 1.79, 2.01, 2.21, 2.42] #+-0.01V
code = [0b111, 0b11000, 0b101101, 0b1000010, 0b1011000, 0b1101011, 0b1111110, 0b10010101, 0b10100111, 0b10111011, 0b11010001, 0b11100101, 0b11111010]
```

```python
plt.scatter(feszultsegek, code)
plt.xlabel("Feszültség(V)")
plt.ylabel("Kijelzett érték")
plt.show()
```

```python
def linearis(x, a, b):
    return x * a + b
```

```python
illesztes = curve_fit(linearis, feszultsegek, code)
print(illesztes[0])
print(np.sqrt(np.diag(illesztes[1])))
```
