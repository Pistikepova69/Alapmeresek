```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
sebesseg_frekvenciak = np.array([13.1, 18.8, 24.9, 31.7, 37.2])
sebesseg_frekvencia_hibak = np.array([0.2, 0.1, 0.2, 0.1, 0.1])
n = np.array([2, 3, 4, 5, 6])
l = 1.532
terjedesi_sebesseg_frekvenciak = np.array([[43.6,	44.1,	43.9],
[50.9,	50.1,	49.9],
[57.2,	57.0,	57.4],
[64.8,	64.7,	64.9],
[71.8,	69.6,	70.6],
[78.2,	77.5,	78.4],
[84.8,	83.5,	85.5]])
atlag_frekvenciak = np.array([np.average(i) for i in terjedesi_sebesseg_frekvenciak])
tomegek = np.array([50, 70, 90, 110, 130, 150, 170]) / 1000
referencia_hossz = 4
referencia_tomeg = 0.8
csomopontok_szama = 2
g = 9.81
```

```python
plt.plot(n, sebesseg_frekvenciak, "D-")
```

```python
plt.plot(tomegek, atlag_frekvenciak ** 2, "D-")
```

```python
def linearis(x, m, b):
    return x * m + b
```

```python
suruseg_illesztes = curve_fit(linearis, tomegek, atlag_frekvenciak ** 2)
```

```python
print(suruseg_illesztes[0])
```

```python
plt.plot(tomegek, atlag_frekvenciak ** 2, ".")
x = np.linspace(50,170, 100) / 1000
plt.plot(x, linearis(x, suruseg_illesztes[0][0], suruseg_illesztes[0][1]))
```

```python
print(9*g/(4*l*l*suruseg_illesztes[0][0]))
print(referencia_tomeg/ referencia_hossz / 1000)
```
